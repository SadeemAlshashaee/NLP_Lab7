# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False) #let the generator complete the scentence, based on probabilties in do sample. not randomly

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt")) #toknize it then return tensors like vectors so the model could understand it

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english") #pipeline, help to upload model, task "sentiment analysis"

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

````python
# 1. Sentiment classification - Encoder (to understand the text)
classifier1 = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
text1 = "The world is enormous, you deserve to explore it"
classifier1(text1)
````

````python
#2. Machine translation - Encoder-Decoder (to understand, then generate in another language)
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
tokenizer1 = AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")
model1 = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M")

text2 = "Translate to Arabic: The world is enormous, you deserve to explore it"

output1 = model1.generate(**tokenizer1(text2, return_tensors="pt"))
print(tokenizer1.decode(output1[0], skip_special_tokens=True))
````

````python
# 3. Open-ended text generation - Decoder

generator1 = pipeline("text-generation", model="distilgpt2")
generator2 = pipeline("text-generation", model="gpt2")
prompt1 = "Sadeem will be a "

output2 = generator1(prompt1, max_new_tokens=30, do_sample=False)
output3 = generator2(prompt1, max_new_tokens=30, do_sample=False)

print(output2[0]["generated_text"])
print(output3[0]["generated_text"])
````

````python
# 4. NER - Encoder (to understand and classify each word)
ner = pipeline("ner", model="dslim/bert-base-NER")

text4 = "Elon Musk founded SpaceX in California."

ner(text4)
````

````python
#5. Text summarization - Encoder-Decoder (to understand, then generate a summary)
tokenizer2 = AutoTokenizer.from_pretrained("google/flan-t5-small")
model2 = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

text5 = """Artificial intelligence is transforming many industries.
It is being used in healthcare, education, transportation, and business.
AI systems can analyze large amounts of data and help people make better decisions."""

input5 = "Summarize: " + text5

output4 = model2.generate(**tokenizer2(input5, return_tensors="pt"),max_new_tokens=50)

print(tokenizer2.decode(output4[0], skip_special_tokens=True))
````
